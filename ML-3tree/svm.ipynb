```python
import pandas as pd
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_classification
```

```python
X,y = make_classification(n_samples=1000, n_features=2, n_classes=2, n_clusters_per_class=2, n_redundant=0)
```

```python
X.shape, y.shape
```

```python
y
```

```python
X = pd.DataFrame(X, columns=['f1', 'f2'])
```

```python
sns.scatterplot(x = X["f1"], y=X["f2"], hue=y)
```

```python
from sklearn.model_selection import train_test_split
```

```python
 X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)
```

```python
from sklearn.svm import SVC
```

```python
classifier = SVC(kernel='linear')
```

```python
classifier.fit(X_train, y_train)
```

```python
classifier.coef_
```

```python
y_pred = classifier.predict(X_test)
```

```python
y_pred
```

## Hyperparameter tuning

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    "C": [0.1,1,10,100,1000],
    "gamma": [1,0.1,0.01,0.001,0.0001]
}
```

```python
grid = GridSearchCV(SVC(), cv=5, param_grid=param_grid, refit=True)
```

```python
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
y_pred_hyper = grid.predict(X_test)
print(y_pred)
print(y_pred_hyper)
```

```python
from sklearn.metrics import accuracy_score, classification_report
```

```python
print(accuracy_score(y_pred, y_test))
print(accuracy_score(y_pred_hyper, y_test))

print(classification_report(y_pred, y_test))
print(classification_report(y_pred_hyper, y_test))
```
